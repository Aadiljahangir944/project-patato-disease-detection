```python
import tensorflow as tf
```

```python
daily_sales_number= [21,22,-108,31,-1,32,34,31]
```

```python
tf_dataset = tf.data.Dataset.from_tensor_slices(daily_sales_number)
tf_dataset
```

```python
for sales in tf_dataset.as_numpy_iterator():
    print(sales)
```

```python
tf_dataset = tf_dataset.filter(lambda x:x>0)
for sales in tf_dataset.as_numpy_iterator():
    print(sales)
```

```python
tf_dataset = tf_dataset.map(lambda x:x*72)
for sales in tf_dataset.as_numpy_iterator():
    print(sales)
```

```python
tf_dataset = tf_dataset.shuffle(2)
for sales in tf_dataset.as_numpy_iterator():
    print(sales)
```

```python
for sale_batch in tf_dataset.batch(3):
    print(sale_batch.numpy())
```

```python
tf_dataset = tf.data.Dataset.from_tensor_slices(daily_sales_number)
tf_dataset = tf_dataset.filter(lambda x:x>0).map(lambda y:y*7).shuffle(2).batch(2)
for sales in tf_dataset.as_numpy_iterator():
    print(sales)
```
